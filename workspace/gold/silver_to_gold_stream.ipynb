```python
stream_df = spark.table("realtimedatabrickhybrid.realtime_pipeline_silver.stream_orders")
```

```python
stream_df.display()
```

```python
from pyspark.sql.functions import count

orders_per_city_df = (
    stream_df
    .groupBy("city")
    .agg(count("order_id").alias("total_orders"))
)
orders_per_city_df.display()
```

```python
from pyspark.sql.functions import sum

revenue_per_city_df = (
    stream_df
    .groupBy("city")
    .agg(sum("amount").alias("total_revenue"))
)

revenue_per_city_df.display()
```

```python
summary_df = stream_df.agg(
    count("order_id").alias("total_orders"),
    sum("amount").alias("total_revenue")
)
summary_df.display()
```

```python
orders_per_city_df.write.mode("overwrite").saveAsTable("realtimedatabrickhybrid.realtime_pipeline_gold.orders_per_city")

revenue_per_city_df.write.mode("overwrite").saveAsTable("realtimedatabrickhybrid.realtime_pipeline_gold.revenue_per_city")

summary_df.write.mode("overwrite").saveAsTable("realtimedatabrickhybrid.realtime_pipeline_gold.realtime_summary")
```
